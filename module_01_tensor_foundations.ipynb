Task 1

```python

import torch
import numpy as np

print("PyTorch:", torch.version)
print("NumPy:", np.version)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
  print("GPU:", torch.cuda.get_device_name(0))
```

```python
#creating 1D tensor
tensor_1d = torch.tensor([1,2,3,4,5,6])
print(tensor_1d)

#creating 2D tensor
tensor_2d = torch.tensor([[1,2,3], [3,4,5], [5,6,7]])
print(tensor_2d)

#creating 3d tensor
tensor_3d = torch.tensor([[[1],[2],[3]],[[4],[5],[6]]])
print(tensor_3d)
```

```python
#get shape
print(tensor_3d.shape)
#get dimension
print(tensor_3d.ndim)
print(tensor_2d.ndim)
#get number of elements
print(tensor_2d.numel())
```

```python
tensor_slice = tensor_2d[1:2, 1:2]
print(tensor_slice)
```
